```python
### 5
# Checking whether the test is passed
# Test is passed when the number of correctly completed
# tasks is at least 50%
#
total_tasks = 20
tasks_ok = int(input("Ile zadan bylo ok?:"))
test_passed = False

if tasks_ok >= (total_tasks)/2:
    test_passed = True

if test_passed:
    print('Congratulations! You passed the test.')
else:
    print('Unfortunately, you failed the test.')
```

```python
### 6
# Checking whether the number
# entered from the keyboard is even or odd 
#
number = int(input('Enter number: '))

if number % 2 == 0:
    print(f'{number} is even')
else:
    print(f'{number} is odd')
```

```python
### 7.1
# Program that calculates the employee's salary,
# taking into account the possibility of receiving a bonus.
#
# The employee does not receive a bonus
basic_salary = 5000
total_salary = 0
bonus = 0.15 # 15%
is_bonus = input('Does the employee receive a bonus? (Y/N):') == 'Y'

if is_bonus:
    total_salary = basic_salary + basic_salary*bonus
else:
    total_salary = basic_salary

print(f'Basic salary: {basic_salary}')
print(f'Bonus included? {is_bonus}')
print(f'Total salary: {total_salary}')
```

```python
### 7.2
# Program that calculates the employee's salary,
# taking into account the possibility of receiving a bonus.
#
# The employee receives a bonus of 30%
basic_salary = 5000
total_salary = 0
bonus = 0.30 # 30%
is_bonus = input('Does the employee receive a bonus? (Y/N):') == 'Y'

if is_bonus:
    total_salary = basic_salary + basic_salary*bonus
else:
    total_salary = basic_salary

print(f'Basic salary: {basic_salary}')
print(f'Bonus included? {is_bonus}')
print(f'Total salary: {total_salary}')
```
